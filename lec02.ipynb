```python
lst = [1,2,3,'hello','c',[2,4]]

print(lst)

#append - is used to add values in the last

lst.append(3)
print(lst)


#extend()- ye puri new list nhi add krta rather unhe elements bna kr add krta hai
lst.extend([2,3,4])
print(lst)

#insert()

lst.insert(2,"value")
print(lst)
```

```python
lst2 = [19,'puja',46,'b',24,'hey',[3,2]]

print(lst2)

lst2.append(5)
print(lst2)

#how to delte from a list
#pop()
lst2.pop()
print(lst2)

#remove()

lst2.pop(2)
print(lst2)

#index

print(lst2.index('puja'))


#count
print(lst.count(46))

```
