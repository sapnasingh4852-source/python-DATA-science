```python
import numpy as np
```

```python
# create a 5*5 matrix of random  integers(1-90)
matrix = np.random.randint(1, 91, size=(5, 5))
print(matrix)
```

```python
# find mean of vaues row and colum wise

row_mean = np.mean(matrix, axis = 1)
col_mean = np.mean(matrix, axis = 0)

print ("row wise mean", row_mean)
print ("colum wise mean",col_mean)
```

```python
# replace all even numbers with -1

matrix [matrix % 2 == 0] = -1
print(matrix)
```

```python
# sort the array row and column wise

sorted_matrix = np.sort(matrix, axis = 1)
print(sorted_matrix)

sorted_matrix = np.sort(matrix, axis = 0)
print(sorted_matrix)
```

```python
#find row and column wise sum separately

row_sum = np.sum(matrix, axis =1)
print(row_sum)

col_sum = np.sum(matrix, axis = 0)
print(col_sum)
```

```python
# count the number of -ve's in each column
count_negatives = np.sum(matrix < 0, axis = 0)
print(count_negatives)
```

```python
# where method to count negative
count_negatives = np.where(matrix < 0, 1, 0)
print(count_negatives)
```
